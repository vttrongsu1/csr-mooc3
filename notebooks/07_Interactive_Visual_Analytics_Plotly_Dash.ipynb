# Interactive Dashboard with Plotly Dash
**IBM Applied Data Science Capstone**

Building reactive dashboards to explore launch success ratios across sites and payload ranges.

```
# Refer to ../scripts/07_spacex_dash_app.py for the full runnable Dash server application.
import pandas as pd
import plotly.express as px

spacex_df = pd.read_csv('../data/spacex_launch_dash.csv')
spacex_df.head()
```

```
# Visualizing Total Success Launches By Site
fig_pie = px.pie(spacex_df, values='class', names='Launch Site', title='Total Successful Launches by All Sites')
fig_pie.show()
```

```
# Visualizing Payload vs. Outcome colored by Booster Version
fig_scatter = px.scatter(
    spacex_df, x='Payload Mass (kg)', y='class',
    color='Booster Version Category',
    title='Correlation Between Payload Mass and Success for All Sites'
)
fig_scatter.show()
```
