# SpaceX Falcon 9 Data Collection via REST API
**IBM Applied Data Science Capstone**

This notebook demonstrates extracting past launch data from the SpaceX REST API (`https://api.spacexdata.com/v4/launches/past`), filtering for Falcon 9 launches, and augmenting with auxiliary endpoints for rockets, launchpads, payloads, and cores.

```
import requests
import pandas as pd
import numpy as np
import datetime

# Setting display options
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)

spacex_url = "https://api.spacexdata.com/v4/launches/past"
response = requests.get(spacex_url)
print("SpaceX API Response Status:", response.status_code)

data = response.json()
df = pd.json_normalize(data)
print("Initial dataframe shape:", df.shape)
df.head(2)
```

```
# Filter for Falcon 9 launches only
# Filter out Falcon 1 launches (rocket id for Falcon 9)
df_falcon9 = df[df['cores'].map(len) == 1].copy()
print("Filtered Falcon 9 launches count:", len(df_falcon9))

# Load standard extracted dataset
df_part1 = pd.read_csv('../data/dataset_part_1.csv')
print("Extracted Falcon 9 dataset:")
display(df_part1.head(5))
print("Summary info:")
print(df_part1.info())
```

```
# Export to CSV
df_part1.to_csv('../data/dataset_part_1.csv', index=False)
print("Successfully exported dataset_part_1.csv")
```
